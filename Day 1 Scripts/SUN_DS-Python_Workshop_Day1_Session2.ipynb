# Tables

In Data Science, a **table is a structured way of organizing data into rows and columns.**

Rows usually represent individual observations or records. Columns represent variables, features, or attributes.





**Tables are a fundamental object type for representing data sets**. A table can be viewed in two ways:

* a sequence of named columns (i.e features or attributes) that each describe a single aspect of all entries in a data set, or

* a sequence of rows (data points or data instances) that each contain all information about a single entry in a data set.....

Tables are important in Data Science because they make data easier to store, inspect, clean, analyze, and model. In Python, tables are commonly represented using a pandas DataFrame.

# Working with tables

To make changes permanent:

```python
# read only specific columns into a new table
totals = olympics[['Country','Total']]
```

```python
# drop a specific column(s)
olympics.drop(columns=['Rank','Country Code','Gold','Silver','Bronze'])
```

**Sorting**

```python
# sort the table according to the values in a specific column
olympics.sort_values(by = "Total")
```

```python
# sort the table according to the values in a specific column in a descending order
olympics.sort_values(by = "Total", ascending=False)
```

You can concatenate commands

```python
# select a two columns and sort according to the values of the one
olympics[['Country','Total']].sort_values(by='Total', ascending=False)
```

**Filtering**

```python
# filter rows\instances according to the values of a specific column
olympics.where(olympics['Gold'] > 10).dropna()
```

```python
# filter rows\instances to retain information with a specific column value between two values
olympics.where(olympics['Gold'].between(2,10)).dropna()
```

```python
# filter rows\instances to retain information with a specific column containing a specific string
olympics.where(olympics['Country'].str.contains('South Africa')).dropna()
```
